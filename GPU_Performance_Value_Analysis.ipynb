```python
from sklearn.preprocessing import MinMaxScaler


# 1. Calculate value metrics
GPU_all['CUDA_per_Dollar'] = (
    GPU_all['CUDA'] / GPU_all['Price_USD']
)

GPU_all['AI_TOPS_per_Dollar'] = (
    GPU_all['AI_TOPS'] / GPU_all['Price_USD']
)

GPU_all['TFLOPS_per_Dollar'] = (
    GPU_all['TFLOPS'] / GPU_all['Price_USD']
)

GPU_all['Bandwidth_per_Dollar'] = (
    GPU_all['Memory_Bandwidth_GB_s'] / GPU_all['Price_USD']
)

GPU_all['Vram_per_Dollar'] = (
    GPU_all['VRAM_GB'] / GPU_all['Price_USD']
)


# 2. Normalize all value metrics
value_columns = [
    'CUDA_per_Dollar',
    'AI_TOPS_per_Dollar',
    'TFLOPS_per_Dollar',
    'Bandwidth_per_Dollar',
    'Vram_per_Dollar'
]

scaler = MinMaxScaler()

value_scaled = scaler.fit_transform(
    GPU_all[value_columns]
)


# 3. Overall Value Score
GPU_all['Value_Score'] = value_scaled.mean(axis=1)


# 4. Create normalized individual scores
GPU_all['CUDA_Value_Score'] = value_scaled[:, 0]

GPU_all['AI_Value_Score'] = value_scaled[:, 1]

GPU_all['TFLOPS_Value_Score'] = value_scaled[:, 2]

GPU_all['Bandwidth_Value_Score'] = value_scaled[:, 3]

GPU_all['VRAM_Value_Score'] = value_scaled[:, 4]


# 5. User input
year = input(
    "Enter release year (2023 / 2024 / 2025): "
)

budget = float(
    input("Enter your budget in USD: ")
)

priority = input(
    "Choose priority (CUDA / AI / TFLOPS / Bandwidth / VRAM / Balanced): "
)


# 6. Filter by year and budget
recommendations = GPU_all[
    (GPU_all['Release_Group'] == year) &
    (GPU_all['Price_USD'] <= budget)
].copy()


# 7. Check if anything fits
if len(recommendations) == 0:

    print("No GPU fits your budget.")

else:

    # 8. Budget fit
    recommendations['Budget_Fit'] = (
        recommendations['Price_USD'] / budget
    )


    # 9. Choose priority score
    if priority == 'CUDA':

        recommendations['Priority_Score'] = (
            recommendations['CUDA_Value_Score']
        )

    elif priority == 'AI':

        recommendations['Priority_Score'] = (
            recommendations['AI_Value_Score']
        )

    elif priority == 'TFLOPS':

        recommendations['Priority_Score'] = (
            recommendations['TFLOPS_Value_Score']
        )

    elif priority == 'Bandwidth':

        recommendations['Priority_Score'] = (
            recommendations['Bandwidth_Value_Score']
        )

    elif priority == 'VRAM':

        recommendations['Priority_Score'] = (
            recommendations['VRAM_Value_Score']
        )

    else:

        recommendations['Priority_Score'] = (
            recommendations['Value_Score']
        )


    # 10. Final recommendation score
    recommendations['Recommendation_Score'] = (
        recommendations['Budget_Fit'] * 0.60
        +
        recommendations['Priority_Score'] * 0.25
        +
        recommendations['Value_Score'] * 0.15
    )


    # 11. Sort
    recommendations = recommendations.sort_values(
        by='Recommendation_Score',
        ascending=False
    )


    # 12. Show top 3
    print("Top 3 Recommended GPUs:")

    display(
        recommendations[
            [
                'Model',
                'Release_Group',
                'Price_USD',
                'Budget_Fit',
                'Priority_Score',
                'Value_Score',
                'Recommendation_Score'
            ]
        ].head(3)
    )
```
